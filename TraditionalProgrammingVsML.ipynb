```python
# ==========================================
# NAME: Jalaludheen
# PRN: 1302250261
# Assignment: Traditional Programming (Loan Approval)
# ==========================================

import pandas as pd


def traditional_loan_system():
  # Read input directly from your Desktop file path
  df = pd.read_excel(r'C:\Users\ecl.lab\Desktop\INPUT.xlsx')

  decisions = []
  for index, row in df.iterrows():
    income = row['Income']
    credit_score = row['Credit_Score']
    existing_loan = row['Existing_Loan']

    # Explicit rules defined by human logic
    if income >= 50000 and credit_score >= 700 and existing_loan <= 20000:
      decision = 'Loan Approved'
    else:
      decision = 'Loan Rejected'
    decisions.append(decision)

  df['Traditional_Decision'] = decisions
  print('--- Traditional Programming Results ---')
  print(
      df[['Income', 'Credit_Score', 'Existing_Loan', 'Traditional_Decision']]
  )
  return df


if __name__ == '__main__':
  traditional_loan_system()
```

```python
# ==========================================
# NAME: Jalaludheen
# PRN: 1302250261
# Assignment: Machine Learning Programming (Loan Approval)
# ==========================================

import pandas as pd
from sklearn.tree import DecisionTreeClassifier


def machine_learning_loan_system():
  # Read input directly from your Desktop file path
  df = pd.read_excel(r'C:\Users\ecl.lab\Desktop\INPUT.xlsx')

  # Features (X) and Target Label (y) learned from data
  X = df[['Income', 'Credit_Score', 'Existing_Loan']]
  y = df['Actual_Approval']  # 1 = Approved, 0 = Rejected

  # Create and train the ML model
  model = DecisionTreeClassifier()
  model.fit(X, y)

  # Predict outputs using learned patterns
  df['ML_Prediction'] = model.predict(X)
  df['ML_Prediction'] = df['ML_Prediction'].map(
      {1: 'Loan Approved', 0: 'Loan Rejected'}
  )

  print('\n--- Machine Learning Results ---')
  print(df[['Income', 'Credit_Score', 'Existing_Loan', 'ML_Prediction']])
  return df


if __name__ == '__main__':
  machine_learning_loan_system()
```
