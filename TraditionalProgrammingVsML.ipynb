```python
# Machine Learning - Loan Approval

from sklearn.tree import DecisionTreeClassifier

# Training data
X = [
    [60000, 750, 10000],
    [30000, 600, 25000],
    [80000, 780, 5000],
    [35000, 620, 30000],
    [70000, 720, 15000],
    [40000, 650, 25000]
]

# 1 = Approved, 0 = Rejected
y = [1, 0, 1, 0, 1, 0]

# Create ML model
model = DecisionTreeClassifier()

# Train the model
model.fit(X, y)

# New customer
new_customer = [[35000, 410, 52000]]

# Prediction
prediction = model.predict(new_customer)

if prediction[0] == 1:
    print("Loan Approved")
else:
    print("Loan Rejected")
```
